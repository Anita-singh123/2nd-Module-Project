# **Project Summary -**

This project focuses on analyzing customer financial data to understand the factors that influence credit scores and to identify patterns associated with Good, Standard, and Poor credit ratings. The dataset contains 100,000 records and 28 variables, covering important customer information such as age, annual income, number of bank accounts, credit cards, loans, delayed payments, outstanding debt, credit utilization ratio, credit inquiries, payment behaviour, and credit score. The main objective of the project is to perform exploratory data analysis (EDA), clean and prepare the data, discover meaningful relationships between variables, and generate insights that can help understand customer creditworthiness.

The data preparation process involved checking the dataset structure, data types, missing values, duplicate records, unique values, and possible inconsistencies. The dataset did not contain complete duplicate rows or pandas missing values. However, some categorical variables contained values such as “NM” (Not Mentioned) and “No Data”, which required appropriate interpretation during analysis. Columns such as ID, SSN, and Name were considered less useful for analytical purposes because they are identification-related rather than meaningful financial predictors. Numerical variables were also examined for unusual values and potential outliers.

Several new features and categories can be created to make the dataset more suitable for analysis. For example, loan information can be summarized using a Loan Count, while financial ratios such as Debt-to-Income Ratio and EMI-to-Income Ratio can provide additional information about a customer's financial burden. Credit history can also be converted into years for easier interpretation. These transformations help improve the understanding of customer financial behaviour.

A Correlation Heatmap was used to understand relationships between numerical variables. The analysis showed that Number of Loans and Outstanding Debt have a relatively strong positive correlation, while Credit Inquiries and Outstanding Debt also show a positive relationship. This indicates that customers with more loans or credit inquiries tend to have higher outstanding debt.

Overall, the project provides a detailed understanding of customer financial characteristics and their relationship with credit scores. The analysis demonstrates how data cleaning, feature engineering, exploratory data analysis, correlation analysis, and visualization can be used together to extract useful insights from financial data.

 Chart - 13 - Correlation Heatmap

````python
# Correlation Heatmap visualization code
numerical_cols = ['Age','Annual_Income','Num_of_Loan', 'Num_of_Delayed_Payment', 'Changed_Credit_Limit', 'Num_Credit_Inquiries', 'Outstanding_Debt']
corr_data = df[numerical_cols].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(corr_data, annot=True, fmt='.2f', cmap='coolwarm', center=0,
            square=True, linewidths=1, cbar_kws={"shrink": 0.8})
plt.title('Correlation Heatmap: Numerical Variables', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()

print("Strongest positive correlations:")
corr_flat = corr_data.unstack().sort_values(ascending=False)
corr_flat = corr_flat[corr_flat < 1.0]  # Exclude self-correlation
print(corr_flat.head(5))
````
